```python
#Exact Missing Years
for subdivision in data['SUBDIVISION'].unique():
    years = set(data[data['SUBDIVISION'] == subdivision]['YEAR'])
    expected_years = set(range(1901, 2016))
    missing_years = sorted(expected_years - years)

    if missing_years:
        print(subdivision)
        print("Missing years:", missing_years)
        print()
```

```python
region_mean = data.groupby('SUBDIVISION')['ANNUAL'].mean().sort_values()

plt.figure(figsize=(10, 8))

region_mean.plot(kind='barh')

plt.xlabel('Average Annual Rainfall (mm)')
plt.ylabel('Subdivision')
plt.title('Average Annual Rainfall by Meteorological Subdivision')

plt.show()
```

## Summary of Data Understanding

### Dataset Structure
- 4,116 observations
- 36 meteorological subdivisions
- Data covers 1901–2015
- 12 monthly rainfall variables
- 4 seasonal aggregate variables
- 1 annual rainfall variable

### Data Quality
- 24 subdivision-year records are missing.
- 26 rows contain one or more missing rainfall values.
- Missing historical observations are concentrated in a few subdivisions.
- No missing values exist in SUBDIVISION or YEAR.

### Statistical Observations
- Rainfall varies significantly across regions.
- June–September have substantially higher average rainfall than most other months.
- July has the highest average monthly rainfall.
- Annual rainfall ranges from 62.3 mm to 6331.1 mm.
- Extreme rainfall values were identified but have not been removed because they may represent genuine observations.

### Feature Relationships
- ANNUAL rainfall is approximately the sum of the 12 monthly rainfall values.
- Seasonal columns are aggregates of monthly rainfall.
- Therefore, annual and seasonal aggregates should not be used as predictors when predicting the same year's rainfall.

### Modeling Consideration
- SUBDIVISION is an important variable because rainfall patterns differ significantly between regions.
- The temporal nature of the dataset must be considered when creating training and testing data.
- Future preprocessing and feature engineering should avoid data leakage.
