#### Creating Artificial Missing Values

```python
np.random.seed(42)

test_data = data.copy()

evaluation_rows = []

for col in monthly_cols:

    available_indices = test_data.index[
        test_data[col].notna()
    ]

    sample_indices = np.random.choice(
        available_indices,
        size=min(20, len(available_indices)),
        replace=False
    )

    for idx in sample_indices:

        evaluation_rows.append({
            'index': idx,
            'column': col,
            'actual': test_data.loc[idx, col]
        })

        test_data.loc[idx, col] = np.nan
```

#### Linear Interpolation

```python
linear_data = test_data.sort_values(
    ['SUBDIVISION', 'YEAR']
).copy()
```

```python
linear_data[monthly_cols] = (
    linear_data
    .groupby('SUBDIVISION')[monthly_cols]
    .transform(
        lambda x: x.interpolate(method='linear')
    )
)
```

#### Suvdivision-wise Mean

```python
mean_data = test_data.copy()
```

```python
for col in monthly_cols:

    mean_data[col] = (
        mean_data
        .groupby('SUBDIVISION')[col]
        .transform(
            lambda x: x.fillna(x.mean())
        )
    )
```

#### KNN

```python
from sklearn.impute import KNNImputer
```

```python
knn_imputer = KNNImputer(n_neighbors=5)
```

```python
knn_data = test_data.copy()

knn_data[monthly_cols] = knn_imputer.fit_transform(
    knn_data[monthly_cols]
)
```

#### Calculate Imputation Errors

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
def evaluate_imputation(imputed_data, evaluation_rows):

    actual_values = []
    predicted_values = []

    for item in evaluation_rows:

        idx = item['index']
        col = item['column']
        actual = item['actual']
        predicted = imputed_data.loc[idx, col]

        if pd.notna(predicted):

            actual_values.append(actual)
            predicted_values.append(predicted)

    mae = mean_absolute_error(
        actual_values,
        predicted_values
    )

    rmse = np.sqrt(
        mean_squared_error(
            actual_values,
            predicted_values
        )
    )

    return mae, rmse
```

#### Comparing all three Methods

```python
linear_mae, linear_rmse = evaluate_imputation(
    linear_data,
    evaluation_rows
)
print("Linear_mae: ",linear_mae)
print("linear_rmse: ",linear_rmse,"\n")

mean_mae, mean_rmse = evaluate_imputation(
    mean_data,
    evaluation_rows
)
print("mean_mae: ",mean_mae)
print("mean_rmse: ",mean_rmse,"\n")

knn_mae, knn_rmse = evaluate_imputation(
    knn_data,
    evaluation_rows
)
print("knn_mae: ",knn_mae)
print("knn_rmse: ",knn_rmse)
```

```python
final_data = data.copy()
```

#### Fill missing monthly rainfall using subdivision-wise mean

```python
for col in monthly_cols:
    final_data[col] = (
        final_data.groupby('SUBDIVISION')[col]
        .transform(lambda x: x.fillna(x.mean()))
    )
```

```python
final_data[monthly_cols].isna().sum()
```

#### Recalculated seasonal totals

```python
final_data['Jan-Feb'] = (
    final_data['JAN'] +
    final_data['FEB']
)

final_data['Mar-May'] = (
    final_data['MAR'] +
    final_data['APR'] +
    final_data['MAY']
)

final_data['Jun-Sep'] = (
    final_data['JUN'] +
    final_data['JUL'] +
    final_data['AUG'] +
    final_data['SEP']
)

final_data['Oct-Dec'] = (
    final_data['OCT'] +
    final_data['NOV'] +
    final_data['DEC']
)
```

#### Recalculate annual rainfall

```python
final_data['ANNUAL'] = final_data[monthly_cols].sum(axis=1)
```
